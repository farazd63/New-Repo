# W02 ML Task Framing

## 1. My Lane as an ML Task
**Task Type:** Classification

This notebook frames a movie star-rating prediction problem as a machine learning task.

## 2. Target or Proxy
Target: `high_rating`

- 1 = Rating >= 4 stars
- 0 = Rating < 4 stars

## 3. Success Metric
F1-score is used because it balances precision and recall.

```
import pandas as pd

df = pd.DataFrame({
    'review_id':[1,2,3,4,5],
    'review_text':['Great','Average','Excellent','Poor','Good'],
    'stars':[5,3,5,2,4]
})
df['high_rating']=(df['stars']>=4).astype(int)
print('Shape:',df.shape)
df.head()
```

## 4. Unit of Analysis
Each row represents **one customer review**.

```
df[['stars','high_rating']].head()
```

## 5. Why ML Beats a Fixed Rule
A fixed rule cannot learn complex patterns from review text, user behavior, and product context. A machine learning model can learn these relationships and generalize to unseen reviews.

## 6. Content Action
If a review is predicted to receive a high rating, it can be highlighted or recommended. Otherwise, it may be flagged for further analysis.
